```python
import os
import pandas as pd

# Configuración de rutas (basada en tu estructura de proyecto)
DATA_DIR = "data"
DATASETS = ["braaksc", "ceradsc", "cogdx"]

def analizar_muestras():
    print(" ANÁLISIS DE MUESTRAS POR CLASE")
    print("-" * 40)
    
    if not os.path.exists(DATA_DIR):
        print(f" Error: El directorio '{DATA_DIR}' no existe.")
        return

    for ds_name in DATASETS:
        file_path = os.path.join(DATA_DIR, f"genes-{ds_name}.csv")
        
        if os.path.exists(file_path):
            try:
                # Cargar el dataset
                df = pd.read_csv(file_path)
                
                # Obtener la última columna (target)
                y = df.iloc[:, -1]
                
                # Contar muestras
                counts = y.value_counts().sort_index()
                total = len(y)
                
                print(f"\nDataset: {ds_name.upper()}")
                print(f"Archivo: {file_path}")
                print(f"Total de muestras: {total}")
                print("Distribución:")
                for clase, count in counts.items():
                    porcentaje = (count / total) * 100
                    print(f"  - Clase '{clase}': {count} muestras ({porcentaje:.2f}%)")
                
            except Exception as e:
                print(f" Error al leer {ds_name}: {e}")
        else:
            print(f" Archivo no encontrado para: {ds_name} ({file_path})")

if __name__ == "__main__":
    analizar_muestras()
```
